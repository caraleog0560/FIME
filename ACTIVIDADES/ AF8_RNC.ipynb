**1.- Cargar librerías.**













```python
import tensorflow as tf
from tensorflow.keras import datasets
import numpy as np
from tensorflow.keras import layers, models
```

**2.- Cargar datos.**

```python
# Cargar el conjunto de datos MNIST
(train_data, train_labels), (test_data, test_labels) = datasets.mnist.load_data()

# Verificar el tamaño de los datos cargados
print(f"Imágenes de entrenamiento: {train_data.shape}")
print(f"Imágenes de prueba: {test_data.shape}")
```

**3.- Preprocesamiento de datos.**

```python
# 1. Normalización: Escalar los valores de los píxeles
train_data = train_data / 255.0
test_data = test_data / 255.0

# 2. Reshape: Ajustar las dimensiones para incluir el canal de color
train_data = train_data.reshape((train_labels.shape[0], 28, 28, 1))
test_data = test_data.reshape((test_labels.shape[0], 28, 28, 1))

print("Datos normalizados y con dimensiones ajustadas.")
print(f"Nueva forma de entrenamiento: {train_data.shape}")
```

**4.- Selección y entrenamiento del modelo.**

```python
# Definición del modelo
model = models.Sequential([
    # Capas Convolucionales para extraer patrones de las imágenes
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Capa para convertir los mapas de características 2D en un vector 1D
    layers.Flatten(),

    # Capas Densas (Prealimentadas) para la clasificación
    layers.Dense(64, activation='relu'),

    # Capa de salida con Softmax (como pide la instrucción) para 10 clases (0-9)
    layers.Dense(10, activation='softmax')
])

# Resumen de la estructura
model.summary()
```

```python
# 1. Compilación
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# 2. Entrenamiento del modelo
print("Iniciando el entrenamiento...")
history = model.fit(train_data, train_labels, epochs=5, batch_size=64)
```

**5.- Prueba del modelo.**

```python
# 1. Evaluación global
print("\nEvaluando el modelo en el conjunto de prueba...")
test_loss, test_acc = model.evaluate(test_data, test_labels, verbose=2)
print(f'Exactitud final: {test_acc:.4f}')
```

```python
# 2. Realizar 5 predicciones
import numpy as np
predicciones = model.predict(test_data[:5])

print("\nResultados de las predicciones:")
for i in range(5):
    valor_predicho = np.argmax(predicciones[i])
    valor_real = test_labels[i]
    print(f"Muestra {i+1}: Predicción -> {valor_predicho} | Valor Real -> {valor_real}")
```
